```python
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_core.documents import Document

# Sample Legal Documents (Pakistani Women Legal Rights / Penal Code excerpts)
legal_docs = [
    Document(
        page_content="Under the Protection Against Harassment of Women at the Workplace Act 2010, every organization must form a three-member Inquiry Committee to investigate workplace harassment complaints.",
        metadata={"source": "Workplace_Act_2010", "section": "Section 3"}
    ),
    Document(
        page_content="Under Section 498A of the Pakistan Penal Code (PPC), depriving a woman of her inherited property through illegal or deceitful means is punishable with imprisonment up to 10 years or a fine of 1 million PKR.",
        metadata={"source": "PPC_498A", "section": "Inheritance Rights"}
    ),
    Document(
        page_content="Under the Muslim Family Laws Ordinance 1961, a wife is entitled to maintenance from her husband during the subsistence of marriage and during the iddat period after divorce.",
        metadata={"source": "MFLO_1961", "section": "Maintenance Rights"}
    )
]

# Free open-source embedding model setup
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Vector Database create karein
vectorstore = Chroma.from_documents(legal_docs, embeddings)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

print("✅ Legal Knowledge Base & Vector Indexing Successfully Created!")
```

```python
def query_legal_agent(user_query):
    print(f"--- 🔍 User Query: {user_query} ---")
    results = retriever.invoke(user_query)

    print("\n⚖️ Relevant Legal Provisions Found:")
    for idx, doc in enumerate(results, 1):
        print(f"\n[Source {idx}]: {doc.metadata['source']} ({doc.metadata['section']})")
        print(f"Details: {doc.page_content}")

# Test Query Run
query_legal_agent("What are the laws regarding women inheritance rights in Pakistan?")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Interactive UI Elements
input_box = widgets.Text(placeholder='Apna legal question type karein...', layout=widgets.Layout(width='70%'))
button = widgets.Button(description='Ask Agent ⚖️', button_style='primary')
output = widgets.Output()

def on_click(b):
    with output:
        clear_output()
        user_query = input_box.value
        if user_query.strip() != "":
            query_legal_agent(user_query)
        else:
            print("Kripya koi question type karein.")

button.on_click(on_click)

print("--- 🤖 LegalMate AI Agent Interactive UI ---")
display(widgets.HBox([input_box, button]))
display(output)
```
